```python
from langgraph.graph import StateGraph, START , END
# from langchain_openai import ChatOpenAI
from langchain_groq import ChatGroq
from typing import TypedDict
from dotenv import load_dotenv
import os
```

```python
GROQ_API_KEY = os.getenv("GROQ_API_KEY")

model = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0,
    groq_api_key="your API key"
)
print(model)
```

```python
# create a state 

class LLM_State(TypedDict):
    question: str
    answer : str 
```

```python
def llm_qa1_fn(state : LLM_State) -> LLM_State:

    # extract question from state 
    question = state['question']

    # form a prompt 
    prompt = f'Answer the following question {question}'

    #ask the question to the LLM
    answer = model.invoke(prompt).content
    # answer = model.invoke(prompt)

    # update the answer in state 
    state['answer'] = answer

    return state
```

```python
# create our graph
graph = StateGraph(LLM_State)

# add nodes 
#node 1 : NODE name, function name
graph.add_node('llm_QA', llm_qa1_fn)


# add edges
graph.add_edge(START, 'llm_QA')
graph.add_edge('llm_QA', END)

#COMPILE
llm_workflow = graph.compile()
```

```python
input_ques = {'question' : 'what is the capital of india'}

output = llm_workflow.invoke(input_ques)

print(output['answer'])
```
